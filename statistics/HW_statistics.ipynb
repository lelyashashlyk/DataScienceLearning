## Первый способ
где я оставила данные в пределах двух нормальных распределений от среднего

```python
new_sal = salaries[(salaries >= np.mean(salaries)-2*np.std(salaries)) & (salaries <= np.mean(salaries)+2*np.std(salaries))]
new_sal_perc = np.round((new_sal.shape[0]/salaries.shape[0])*100, 2) # чтобы найти процентное соотношение ген.выборки и выборки без выбросов

display(new_sal.min(), new_sal.max())

fig, ax = plt.subplots(1, 3, figsize=(10, 5))

# рисуем диаграмку
ax[0].pie([new_sal_perc, 100 - new_sal_perc], autopct = '%.2f%%',
          startangle=90, colors=['lightpink', 'mistyrose'], wedgeprops=dict(edgecolor='black', linewidth=0.6))
ax[0].set_title('Процентное отношение долей')

sns.boxplot(salaries, ax=ax[1], color = '#6f49f5',boxprops=dict(edgecolor='black', linewidth=0.8),
    whiskerprops=dict(color='black', linewidth=0.8),
    capprops=dict(color='black', linewidth=0.8),
    medianprops=dict(color='black', linewidth=0.8),
    flierprops=dict(marker='o', color='black', markersize=7)).set_title('General sample')
sns.boxplot(new_sal, color = '#a188f7',
    boxprops=dict(edgecolor='black', linewidth=0.8),
    whiskerprops=dict(color='black', linewidth=0.8),
    capprops=dict(color='black', linewidth=0.8),
    medianprops=dict(color='black', linewidth=0.8)).set_title('New sample')

plt.subplots_adjust(wspace=0.5)
plt.show()
```

## Второй способ
где я фильтрую выборку через 1.5 квартильных размаха от квартилей


```python
q1 = np.quantile(salaries, 0.25)
q3 = np.quantile(salaries, 0.75)

new_sal_2 = salaries[(salaries >= q1 - 1.5*(q3-1)) & (salaries <= q3 + 1.5*(q3-q1))]
new_sal_perc_2 = np.round((new_sal_2.shape[0]/salaries.shape[0])*100, 2)

display(new_sal_2.min(), new_sal_2.max())

fig, ax = plt.subplots(1, 3, figsize=(10, 5))

# рисуем диаграмку
ax[0].pie([new_sal_perc_2, 100 - new_sal_perc], autopct = '%.2f%%',
          startangle=90, colors=['#b2db42', '#caf53d'], wedgeprops=dict(edgecolor='black', linewidth=0.6))
ax[0].set_title('Процентное отношение долей')

sns.boxplot(salaries, ax=ax[1], color = '#7dd11d',boxprops=dict(edgecolor='black', linewidth=0.8),
    whiskerprops=dict(color='black', linewidth=0.8),
    capprops=dict(color='black', linewidth=0.8),
    medianprops=dict(color='black', linewidth=0.8),
    flierprops=dict(marker='o', color='black', markersize=7)).set_title('General sample')
sns.boxplot(new_sal_2, color = '#619606',
    boxprops=dict(edgecolor='black', linewidth=0.8),
    whiskerprops=dict(color='black', linewidth=0.8),
    capprops=dict(color='black', linewidth=0.8),
    medianprops=dict(color='black', linewidth=0.8)).set_title('New sample')

plt.subplots_adjust(wspace=0.5)
plt.show()
```
