```python
diamonds.head()
```

```python
# построить барплот с количеством алмазов разной чистоты для премиум качества
# Как я поняла, премиум качество - это бриллианты с огранкой "Премиум"

sns.barplot(diamonds.loc[diamonds.cut=='Premium', 'clarity'].value_counts())
plt.title("Premium quality's clarity barplot")
plt.show()
```

```python
# отфильтровать строки с ценой выше Q2 и определить самый распространенный среди них цвет. сколько всего камней такого цвета?

q2 = diamonds.price.quantile(0.5)
(diamonds.loc[diamonds.price > q2].value_counts('color').idxmax(),
 diamonds.loc[diamonds.price > q2].value_counts('color').max())
```

```python
# найти среднюю цену и стандартное отклонение цены для алмазов идеального или премиум качества

(round(diamonds.loc[(diamonds.cut=='Premium') | (diamonds.cut=='Ideal'), 'price'].mean().item(), 2),
round(diamonds.loc[(diamonds.cut=='Premium') | (diamonds.cut=='Ideal'), 'price'].std(), 2))
```

```python
# построить боксплоты цены для каждого типа clarity

sns.boxplot(x = diamonds.clarity, y = diamonds.price)
plt.title('Boxplots of price for every clarity')
plt.show()
```

## Task 8
Используя датасет cars, достаньте первое слово из колонки name и соберите их в новую колонку brand. Название бренда должно быть с большой буквы.

```python
def make_brand(s:str)->str:
  return s.split('_')[0].capitalize()
```

```python
cars.name.map(make_brand)
```

```python
cars['brand'] = cars.name.map(make_brand)
```

```python
cars.brand.unique()
```

```python
corr_dict = {'Amc':'GMC',
             'Bmw':'BMW',
             'Chevy':'Chevrolet',
             'Chevroelt':'Chevrolet',
             'Mercedes-benz':'Mercedes-Benz',
             'Toyouta':'Toyota',
             'Vw':'Volkswagen',
             'Maxda':'Mazda'
}
```

```python
cars.brand = cars.brand.replace(corr_dict)
```
